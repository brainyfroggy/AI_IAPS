```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
rois = np.load('./outputs/Sub1.npy')
```

```python
decode_names = []
decode_names.append('pleasant vs. neutral')
decode_names.append('pleasant vs. unpleasant')
decode_names.append('unpleasant vs. neutral')
decode_names.append('pleasant vs. neutral (natrual)')
decode_names.append('pleasant vs. unpleasant (natrual)')
decode_names.append('unpleasant vs. neutral (natrual)')
decode_names.append('pleasant vs. neutral (AI)')
decode_names.append('pleasant vs. unpleasant (AI)')
decode_names.append('unpleasant vs. neutral (AI)')
decode_names.append('AI vs. natrual (pleasant)')
decode_names.append('AI vs. natrual (neutral)')
decode_names.append('AI vs. natrual (unpleasant)')
```

```python
roi_name = [
    'V1v', 
    'V1d', 
    'V2v',
    'V2d',
    'V3v',
    'V3d',
    'hV4', 
    'VO1', 
    'VO2', 
    'PHC1', 
    'PHC2', 
    'MST', 
    'hMT', 
    'LO2', 
    'LO1', 
    'V3b', 
    'V3a', 
    'IPS0', 
    'IPS1', 
    'IPS2', 
    'IPS3', 
    'IPS4', 
    'IPS5', 
    'SPL1', 
    'FEF', 
    'ACCpre', 
    'ACCsub', 
    'ACCsup', 'Amygdala', 'Angular', 'Calcarine', 'Caudate', 'Cuneus', 'Fusiform', 'Hippocampus', 'Insula', 'Lingual', 'ParaHippocampal', 'Postcentral', 'Precentral', 'Precuneus', 'Putamen', 'ThalPuL']
```

```python
# Set the size of the figure
fig = plt.figure(figsize=(15, 20))  # Width: 8 inches, Height: 6 inches

# Plot the heatmap
plt.imshow(rois, cmap='viridis', interpolation='nearest', extent=[0, len(decode_names), 0, len(roi_name)])
# Set x-axis and y-axis labels
plt.xticks(np.arange(len(decode_names)) , decode_names)
plt.yticks(np.arange(len(roi_name)), roi_name)

plt.colorbar()
# Add text annotations on top of each cell
for i in range(len(decode_names)):
    for j in range(len(roi_name)):
        plt.text(j , i , f"{rois[i, j]:.2f}", color='white', ha='center', va='center')

# Automatically adjust layout
plt.tight_layout()
plt.title("Heatmap")
plt.show()
```
