```python
import pandas as pd
import shutil
import os
```

```python
csv_file_path = 'F:\yujun\projects\AI_IAPS\GLM_singletrial\/beta_groups.csv'
dataDir = '../GLM_singletrial/betas/'
subs = os.listdir(dataDir)
subs
# nii_files_dir = 'path/to/your/nii/files'
# output_dir = 'path/to/your/output/directory'
```

```python
subs = ['Sub1', 'Sub2', 'Sub3', 'Sub4', 'Sub5', 'Sub6', 'Sub7']
```

```python
csv_file = pd.read_csv(csv_file_path, header= None)
csv_file
```

```python
for sub in subs:
    sub_dir = os.path.join(dataDir, sub)
    for index, row in csv_file.iterrows():
        nii_file_name = row[0]
        category = row[1]

        # Define the source and destination paths
        source_path = os.path.join(sub_dir, nii_file_name)
        dest_path = os.path.join(sub_dir, category)
        if not os.path.exists(dest_path):
            os.makedirs(dest_path)

        shutil.copy(source_path, dest_path)
```
